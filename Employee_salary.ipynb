```python
import pandas as pd

# 1. Create employee salary data
salary_data = {
    "Employee": [
        "Employee 1",
        "Employee 2",
        "Employee 3",
        "Employee 4",
        "Employee 5"
    ],
    "Salary": [
        300000,
        450000,
        500000,
        600000,
        750000
    ]
}

# 2. Create DataFrame
df = pd.DataFrame(salary_data)

# 3. Calculate salary statistics
total_salary = df["Salary"].sum()
average_salary = df["Salary"].mean()
highest_salary = df["Salary"].max()
lowest_salary = df["Salary"].min()

employee_count = df["Employee"].count()

# 4. Display employee data
print("EMPLOYEE SALARY DATA")
print(df.to_string(index=False))

# 5. Display analysis results
print("\nSALARY ANALYSIS")

print("Number of employees:", employee_count)

print(f"Total Salary: ₹{total_salary:,.2f}")

print(f"Average Salary: ₹{average_salary:,.2f}")

print(f"Highest Salary: ₹{highest_salary:,.2f}")

print(f"Lowest Salary: ₹{lowest_salary:,.2f}")

# 6. Compare salaries with average
df["Difference from Average"] = (
    df["Salary"] - average_salary
)

print("\nCOMPARISON WITH AVERAGE")

print(df.to_string(index=False))

# 7. Save results to CSV
df.to_csv("salary_analysis_results.csv", index=False)

print("\nResults saved successfully!")
```
